```python
selected_cols = ['vehicle_weight', 'model_year']

filtered_df = df.loc[df.origin == "Asia", selected_cols].iloc[:7]

filtered_df
```

```python
X = np.array(filtered_df)
XTX = X.T.dot(X)
```

```python
XTX
```

```python
XTX_INV = np.linalg.inv(XTX)
XTX_INV
```

```python
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
```

```python
w = XTX_INV.dot(X.T).dot(y)
w
```
